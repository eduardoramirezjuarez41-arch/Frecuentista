#$Ejercicio$ 1

Simula 10 000 lanzamientos de dos datos y estima la probabilidad de que la suma sea exactamente 5. Compara esa frecuencia con la probabilidad clásica que ya calculaste.



```python
import random

lanzamientos = 10000
favorables = 0

for i in range(lanzamientos):
    dado1 = random.randint(1, 6)
    dado2 = random.randint(1, 6)

    if dado1 + dado2 == 5:
        favorables += 1

probabilidad = favorables / lanzamientos
probabilidad_clasica = 4 / 36

print("Lanzamientos:", lanzamientos)
print("Casos favorables:", favorables)
print("Probabilidad frecuentista:", probabilidad)
print("Probabilidad clásica:", probabilidad_clasica)
print("Porcentaje frecuentista:", probabilidad * 100, "%")
```

#$Ejercicio$ 2

Simula 10 000 PIN de 4 dígitos, del 0 al 9, y estima la probabilidad de que ninguno se repita. Compara el resultado con la fracción de la parte clásica.

```python
import random

intentos = 10000
favorables = 0

for i in range(intentos):
    pin = []

    for j in range(4):
        pin.append(random.randint(0, 9))

    if len(set(pin)) == 4:
        favorables += 1

probabilidad = favorables / intentos
probabilidad_clasica = (10 * 9 * 8 * 7) / (10 * 10 * 10 * 10)

print("PIN simulados:", intentos)
print("Casos favorables:", favorables)
print("Probabilidad frecuentista:", probabilidad)
print("Probabilidad clásica:", probabilidad_clasica)
print("Porcentaje frecuentista:", probabilidad * 100, "%")
print("Porcentaje clásico:", probabilidad_clasica * 100, "%")
```

#$Ejercicio$ 3

Simula 20.000 lanzamientos de tres dados. Estima la frecuencia de la suma 9 y la de la suma 10. ¿Se nota, en la simulación, cuál de las dos sale más?



```python
import random

def lanzar_dado():
    return random.randint(1, 6)

lanzamientos = 20000
suma_9 = 0
suma_10 = 0

for i in range(lanzamientos):
    dado1 = lanzar_dado()
    dado2 = lanzar_dado()
    dado3 = lanzar_dado()

    suma = dado1 + dado2 + dado3

    if suma == 9:
        suma_9 += 1
    if suma == 10:
        suma_10 += 1

print("Lanzamientos:", lanzamientos)
print("Veces que salió 9:", suma_9)
print("Veces que salió 10:", suma_10)

print("Frecuencia de la suma 9:", suma_9 / lanzamientos)
print("Frecuencia de la suma 10:", suma_10 / lanzamientos)
```

#$Ejercicio$ 4

Basado en el Ejemplo 3, simula una urna con 3 bolas rojas y 2 azules. Realiza 2,000 extracciones (con reemplazo) y calcula la probabilidad de obtener una azul.

```python
import random

urna = ["roja", "roja", "roja", "azul", "azul"]
extracciones = 2000
azules = 0

for i in range(extracciones):
    bola = random.choice(urna)

    if bola == "azul":
        azules += 1

probabilidad = azules / extracciones

print("Extracciones:", extracciones)
print("Bolas azules:", azules)
print("Probabilidad frecuentista:", probabilidad)
print("Porcentaje:", probabilidad * 100, "%")
```

#$Ejercicio$ 5

Una serie son 20 volados. Estima, con 10 000 series, la probabilidad de que aparezcan 5 águilas seguidas en algún momento de la serie.

```python
import random

def lanzar_moneda():
    return random.choice(["águila", "sol"])

series = 10000
favorables = 0

for i in range(series):
    racha = 0

    for j in range(20):
        moneda = lanzar_moneda()

        if moneda == "águila":
            racha += 1
        else:
            racha = 0

    if racha >= 5:
        favorables += 1

probabilidad = favorables / series

print("Series favorables:", favorables)
print("Probabilidad:", probabilidad)
print("Porcentaje:", probabilidad * 100, "%")
```

#$Ejercicio$ 6

Los clientes llegan a una tienda de forma aleatoria (entre 1 y 5 minutos). El cajero tarda entre 2 y 4 minutos en atender. Simula la llegada de 500 clientes y calcula la probabilidad de que un cliente tenga que esperar más de 10 minutos en la fila.

```python
import random

clientes = 500
tiempo_llegada = 0
tiempo_cajero = 0
esperas = 0

for i in range(clientes):
    tiempo_llegada += random.randint(1, 5)
    espera = max(0, tiempo_cajero - tiempo_llegada)

    if espera > 10:
        esperas += 1

    tiempo_cajero = max(tiempo_llegada, tiempo_cajero) + random.randint(2, 4)

probabilidad = esperas / clientes

print("Total de clientes:", clientes)
print("Clientes que esperaron más de 10 minutos:", esperas)
print("Probabilidad estimada:", probabilidad)
print("Porcentaje:", probabilidad * 100, "%")
```
